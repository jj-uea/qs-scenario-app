```python
import pandas as pd
```

```python
df = pd.read_csv('../data/qs_data.csv', encoding='latin1')
```

```python
df["metric"] = df["metric"].replace({
    "International Students": "International Student Ratio",
    "International Faculty": "International Faculty Ratio",
})
```

```python
df["score"] = (
    pd.to_numeric(
        df["score"].replace("-", pd.NA),
        errors="coerce"
    )
)
```
