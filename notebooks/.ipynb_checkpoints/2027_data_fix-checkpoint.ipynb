```python
import pandas as pd
```

```python
df = pd.read_csv('../data/qs_data.csv', encoding='latin1')
```

```python
df
```

```python
# remove international student diversity - no longer included since 2027
df = df[df['metric']!='International Student Diversity']
```

```python
df["metric"] = df["metric"].replace({
    "International Students": "International Student Ratio",
    "International Faculty": "International Faculty Ratio",
})
```

```python
df.head()
```

```python
# Replace any dashes with blank values.
df["score"] = (
    pd.to_numeric(
        df["score"].replace("-", pd.NA),
        errors="coerce"
    )
)
```

```python
# now remove any overalls with a blank value
df = df[~((df["metric"] == "Overall") & (df["score"].isna()))]
```

```python
df.info()
```

```python
df['metric'].unique()
```

```python
df.to_csv('../data/qs_data.csv', index=False)
```
